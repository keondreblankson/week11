For this assignment, I chose the New York Times Most Popular API. The code gets the most viewed articles from the past seven days and puts the information into a pandas DataFrame.

```python
import requests
import pandas as pd
from getpass import getpass
from IPython.display import display
```

```python
api_key = getpass("Enter your NYT API key: ").strip()
if not api_key:
    raise ValueError("Enter your API key before continuing.")
```

```python
url = "https://api.nytimes.com/svc/mostpopular/v2/viewed/7.json"

response = requests.get(url, params={"api-key": api_key}, timeout=30)
print("Status code:", response.status_code)

if response.status_code != 200:
    raise RuntimeError("Request failed. Check your key and that the Most Popular API is enabled.")

data = response.json()
if data.get("status") != "OK":
    raise RuntimeError("The API did not return a successful response.")
```

```python
# Put the articles into a DataFrame
df = pd.DataFrame(data["results"])

# Keep the columns I want to use
columns = ["title", "section", "byline", "published_date", "abstract", "url"]
df = df.reindex(columns=columns).copy()
df["published_date"] = pd.to_datetime(df["published_date"])

print("Total articles:", len(df))
display(df.head())
```

The code uses requests to get the article data and reads the response as JSON. It then takes the articles from the results list and puts them into a pandas DataFrame. I included the title, section, author, date, summary, and link so the information is easier to read.
